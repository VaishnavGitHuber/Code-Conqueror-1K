### 26 INTERSECTION OF 2 SORTED ARRAYS
- https://www.geeksforgeeks.org/problems/intersection-of-two-sorted-array-1587115620/1

```python
def intersection(arr1, arr2):
        intersection = []
         
        i = j = 0
        while i < len(arr1) and j < len(arr2):
            if arr1[i] == arr2[j]:
                if not intersection or intersection[-1] != arr1[i]:
                    intersection.append(arr1[i])
                i += 1
                j += 1
            elif arr1[i] < arr2[j]:
                i += 1
            else:
                j += 1
        return intersection
```

### 27 SUBARRAYS SUM EQUALS K
- https://leetcode.com/problems/subarray-sum-equals-k/description/

```python
### Brute force approach 
def subarraySum(nums, k):
        n = len(nums)
        count = 0

        for i in range(n):
            sum_n = 0
            for j in range(i,n):
                sum_n += nums[j]
                # if sub arrays equals k
                if sum_n == k:
                    count += 1
        return count
```

```python
def subarraySum(nums,k):
        count = 0
        prefix_sum = 0
        prefix_map = {0:1}

        for i in range(len(nums)):
            prefix_sum += nums[i]
            # condition to check sub array
            if prefix_sum - k in prefix_map:
                count += prefix_map[prefix_sum - k]
            prefix_map[prefix_sum] = prefix_map.get(prefix_sum, 0) + 1
        return count
```

### 28 Sort Colors/Sort 0s,1s,2s
- https://leetcode.com/problems/sort-colors/
- https://www.geeksforgeeks.org/problems/sort-an-array-of-0s-1s-and-2s4231/1

```python
# Merge sort 
# T O(nlogn)
# S O(n) 
```

```python
def sort012(arr):
        count0 = 0
        count1 = 0
        count2 = 0
        
        for i in range(len(arr)):# O(n)
            if arr[i] == 0:
                count0 += 1
            elif arr[i] == 1:
                count1 += 1
            else:
                count2 += 1
        
        for i in range(0,count0):# O(n)
            arr[i] = 0
        for i in range(count0, count0+count1):
            arr[i] = 1
        for i in range(count0+count1, count0+count1+count2):
            arr[i] = 2
        return arr
```

- Time complexcity : O(2n)
- Space complecity : O(1)

```python
def sortColors(nums):
        # use the Dutch Netional Flag algorithm 
        low, mid, high = 0,0, len(nums)-1

        while mid <= high:
            if nums[mid] == 0:
                nums[low], nums[mid] = nums[mid], nums[low]
                low += 1
                mid += 1
            elif nums[mid] == 1:
                mid += 1
            else:
                nums[high], nums[mid] = nums[mid], nums[high]
                high -= 1
        return nums
```

- Time complexcity : O(n)
- Space complexcity : O(1)

### 29 Find Missing and Repeated Value 
- https://leetcode.com/problems/find-missing-and-repeated-values/

```python
class Solution:
    def findMissingAndRepeatedValues(self, grid):
        row, col = len(grid), len(grid[0])
        num_count = dict()
        repeated_value = None
        missing_value = None

        for i in range(row):
            for j in range(col):
                if grid[i][j] in num_count:
                    num_count[grid[i][j]] += 1
                else:
                    num_count[grid[i][j]] = 1
        
        for i in range(1, row*col+1):
            if i not in num_count:
                missing_value = i
            else:
                if num_count[i] > 1:
                    repeated_value = i
        return [repeated_value, missing_value]
```

### 30. Maximum Sub Array
- https://leetcode.com/problems/maximum-subarray/

```python
## Brute force approach - Time Limit exceeed error
class Solution:
    def maxSubArray(self, nums):
        n = len(nums)
        sum_max = float('-inf')

        for i in range(n):
            sum_ = 0
            for j in range(i, n):
                sum_ += nums[j]
                sum_max = max(sum_max, sum_)
        return sum_max
```

```python
## Optimised 
class Solution:
    def maxSubArray(self, nums):
        n = len(nums)
        sum_max = float('-inf')
        sum_ = 0
        
        for i in range(n):
            sum_ += nums[i]
            sum_max = max(sum_max, sum_)

            if sum_ < 0:
                sum_ = 0
        return sum_max
```
